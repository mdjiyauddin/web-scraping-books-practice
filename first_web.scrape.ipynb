```python
import requests, time

for i in range(1, 51):
    a = requests.get(f"https://books.toscrape.com/catalogue/page-{i}.html")
    if a.status_code == 200:
        with open(f"htmls/page{i}.html", "w", encoding="utf-8") as f:
            f.write(a.text)
        print(f"Downloaded page {i} successfully")
    else:
        print(f"FAILED page {i} — status {a.status_code}")
    time.sleep(0.5)   # thoda gap do requests ke beech
```

```python
from bs4 import BeautifulSoup

rating_map = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
books_data = []

for i in range(1, 51):
    with open(f"htmls/page{i}.html", "r", encoding="utf-8") as f:
        soup = BeautifulSoup(f, "html.parser")
    books = soup.select("article.product_pod")
    for book in books:
        title = book.select_one("h3 a")["title"]
        price = book.select_one("p.price_color").get_text(strip=True)
        rating = rating_map[book.select_one("p.star-rating")["class"][1]]
        stock = book.select_one("p.instock.availability").get_text(strip=True)
        books_data.append({"title": title, "price": price, "rating": rating, "stock": stock})

top_books = [b for b in books_data if b["rating"] == 5]
best_book = max(books_data, key=lambda b: b["rating"])
long_book = max(books_data, key=lambda b: len(b["title"]))
short_book = min(books_data, key=lambda b: len(b["title"]))

print("="*50)
print(f"TOP RATED BOOKS (5★) — Total: {len(top_books)}")
print("="*50)
for idx, b in enumerate(top_books, start=1):
    print(f"{idx:2}. {b['title']:<60} | {b['price']} | {b['stock']}")

print("\n" + "="*50)
print("SUMMARY")
print("="*50)
print(f"Best Book      : {best_book['title']} (Rating: {best_book['rating']}★)")
print(f"Longest Title  : {long_book['title']}")
print(f"Shortest Title : {short_book['title']}")
```
